#### PART 0

## **UNDERSTANDING** Radar **Charts**

<div align="center" style="font-size:12px; font-family:FreeMono; font-weight: 100; font-stretch:ultra-condensed; line-height: 1.0; color:#2A2C2B">
    <img src="images/several_indiv_radarchart.png" align="center" width="40%" padding="10px"><br>
    <br>
</div>


A radar chart (also called a spider chart or web chart) represents multivariate data by mapping each variable to its own axis radiating from a center point. All axes are equally spaced around 360 degrees and share the same scale. Each observation becomes a closed polygon connecting its value on every axis - the shape of that polygon is the primary signal the chart communicates.

<br>

**When to use a radar chart**

Radar charts communicate well when:

- You have **3 to 6 variables** that either already share a scale or can be meaningfully normalized to one.
- You are comparing **2 to 4 entities** - at this count, overlapping polygons remain distinguishable.
- The **shape** of the polygon carries interpretive meaning. A balanced polygon signals even performance across dimensions; a collapsed side signals a specific weakness. This at-a-glance shape reading is the chart's primary advantage over a table or grouped bar chart.

Common uses: athlete performance profiles, product feature comparisons, survey result summaries, model evaluation across multiple metrics.

<br>

**When to avoid a radar chart**

- **More than 6 variables** - the polygon becomes dense and individual axis values are hard to read. The filled area also loses discriminative value as the polygon approaches a full circle.
- **More than 4 entities** - overlapping filled polygons obscure each other. Color alone cannot separate many overlapping shapes.
- **Variables on different or unconvertible scales** - a variable measured 0-1000 dominates polygon area over one measured 0-1 even after the axes are individually scaled. Area comparisons become misleading.
- **Precise value reading is required** - readers cannot estimate values from polygon vertices as accurately as from bar chart heights.

___

**Note:** Polygon area is a function of both values and angular spacing between axes. The same data produces different polygon areas depending on the axis order - a known perceptual bias. When the axis order matters to your audience's interpretation, label the ordering logic explicitly or use a parallel coordinates plot instead.

___

<br>

**Alternatives**

| Situation | Better choice |
|---|---|
| Many variables (7+) or many entities (5+) | Parallel coordinates plot |
| Precise value comparison is required | Grouped bar chart |
| One entity, many metrics over time | Line chart or heatmap |

Parallel coordinates represent each variable as a vertical axis; individual observations are lines crossing these axes. The layout handles far more dimensions and entities than a radar chart, and preserves the ability to spot correlations between adjacent axes. It can be argued that a radar chart is a radially arranged variant of the parallel coordinates plot.

#### PART 1

## **BASIC** Radar **Chart**

<div align="center" style="font-size:12px; font-family:FreeMono; font-weight: 100; font-stretch:ultra-condensed; line-height: 1.0; color:#2A2C2B">
    <img src="images/radar_plots_many.png" align="center" width="40%" padding="10px"><br>
    <br>
</div>


The simplest way to build a radar chart in Matplotlib is to use the built-in polar coordinate system via `plt.subplot(111, polar=True)`. This switches the axes from Cartesian (x, y) to polar (radius r, angle theta) without any custom code. Distance from the center maps to the data value; the angular position maps to the variable being measured.

<br>

**Computing the angles**

For N variables, divide a full circle ($2\pi$ radians) into N equal slices:

$$\theta_k = \frac{k}{N} \cdot 2\pi \quad \text{for } k = 0, 1, \ldots, N-1$$

where $\theta_k$ is the angle assigned to the $k$-th variable and N is the total number of variables. This places each variable's axis at an equal angular distance from its neighbors.

The last step - `angles += angles[:1]` - appends the first angle to the end of the list. When `ax.plot()` connects consecutive points, this forces the line back to the starting axis, closing the polygon. Without it, the last axis and the first axis leave an open gap in the drawn shape.

<br>

**Orientation controls**

`ax.set_theta_offset(pi / 2)` rotates the entire chart so that angle = 0 (which would normally point right along the positive x-axis) points straight up instead. This is a display convention: most readers expect the first axis to appear at the top of a radar chart, not to the right.

`ax.set_theta_direction(-1)` reverses the angular direction from counterclockwise (the trigonometric default) to clockwise. Combined with the offset, the axes now read top-to-right-to-bottom-to-left, matching the clockwise reading order most users expect.

<br>

**When to choose this approach**

Use the basic polar subplot when a circular frame is acceptable and you want minimal setup. The limitation is that Matplotlib's built-in polar axes always draw a circle as the chart boundary. If you need the boundary to follow a polygon shape whose edges align with the axes (a pentagon for 5 variables, a hexagon for 6), the basic approach cannot do this without the custom projection in Part 2.

<br>

*Sources Consulted:*
- Python Graph Gallery: [Radar Chart with Several Individuals](https://python-graph-gallery.com/391-radar-chart-with-several-individuals/)
- Matplotlib documentation: [Polar plot](https://matplotlib.org/stable/gallery/pie_and_polar_charts/index.html)


````python
df = pd.DataFrame({
    'group': ['A', 'B', 'C', 'D'],
    'var1': [38, 1.5, 30, 4],
    'var2': [29, 10, 9, 34],
    'var3': [8, 39, 23, 24],
    'var4': [7, 31, 33, 14],
    'var5': [28, 15, 32, 14]
})

categories = list(df)[1:]
N = len(categories)

# Compute one angle per variable, equally spaced around 2*pi.
# The extra angles[:1] appended at the end closes the polygon on plot/fill.
angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]

ax = plt.subplot(111, polar=True)
ax.set_theta_offset(pi / 2)    # rotate so first axis points up
ax.set_theta_direction(-1)     # clockwise axis order

plt.xticks(angles[:-1], categories)

ax.set_rlabel_position(0)
plt.yticks([10, 20, 30], ["10", "20", "30"], color="grey", size=7)
plt.ylim(0, 40)

values = df.loc[0].drop('group').values.flatten().tolist()
values += values[:1]
ax.plot(angles, values, linewidth=1, linestyle='solid', label='group A')
ax.fill(angles, values, 'b', alpha=0.1)

values = df.loc[1].drop('group').values.flatten().tolist()
values += values[:1]
ax.plot(angles, values, linewidth=1, linestyle='solid', label='group B')
ax.fill(angles, values, 'r', alpha=0.1)

plt.legend(loc='upper right', bbox_to_anchor=(0.1, 0.1))
plt.savefig('images/radar_plot_basic.png', bbox_inches='tight', facecolor='white')
plt.show()
````

#### PART 2

## **ADVANCED** Radar Chart - **Custom Projection**

<div align="center" style="font-size:12px; font-family:FreeMono; font-weight: 100; font-stretch:ultra-condensed; line-height: 1.0; color:#2A2C2B">
    <img src="images/several_indiv_radarchart.png" align="center" width="40%" padding="10px"><br>
    <br>
</div>


Matplotlib's built-in polar axes always draw a **circle** as the chart boundary. For radar charts it is often more visually coherent to clip the chart to a polygon whose edges align with the axes - a pentagon for 5 variables, a hexagon for 6. Achieving this requires registering a custom projection.

<br>

**Why a custom projection**

`register_projection(RadarAxes)` adds `RadarAxes` to Matplotlib's internal registry of axes types. After registration, `fig.add_subplot(projection="radar")` creates a `RadarAxes` instance instead of the default `Axes`. This is the same mechanism Matplotlib uses for its built-in 3D axes (`projection="3d"`) - a registered name maps to an `Axes` subclass that controls how data is drawn and how the frame looks.

<br>

**How `radar_factory` is structured**

`radar_factory(num_vars, frame)` does three things:

1. **Computes angles** - the same equally spaced fractions of $2\pi$ from Part 1, plus a $\pi/2$ rotation so the first axis points up. These angles are stored in the enclosing function's scope where the patch-drawing functions can access them.

2. **Defines frame shapes** - `draw_poly_patch` returns a `plt.Polygon` built from the axis vertices; `draw_circle_patch` returns a `plt.Circle`. Choosing `frame="polygon"` or `frame="circle"` selects which function becomes `RadarAxes.draw_patch`.

3. **Defines and registers `RadarAxes`** - a subclass of `PolarAxes` with four overridden methods:
   - `plot` and `fill` call `_close_line`, which appends the first data point to the end of each plotted array - the same close trick from Part 1, but applied automatically to every `ax.plot()` and `ax.fill()` call so the caller never needs to do it manually.
   - `_gen_axes_patch` returns the polygon or circle as the axes background clip region.
   - `_gen_axes_spines` overrides the axes border. Instead of the default circle, it constructs a `Spine` object that follows a `Path` built from the polygon vertices. The `Spine` is what clips the drawn content to the polygon boundary - any line or fill outside the polygon path is hidden.

___

**Note:** `Spine` wraps a `Path` and draws it as the visible border of the axes. Overriding `_gen_axes_spines` with a polygon path is what makes the chart content appear clipped inside the polygon rather than inside a circle.

___

<br>

*Sources Consulted:*
- Matplotlib documentation: [Radar chart (aka spider or star chart)](https://matplotlib.org/stable/gallery/specialty_plots/radar_chart.html)
- Matplotlib documentation: [Custom Projections](https://matplotlib.org/stable/gallery/misc/custom_projection.html)


````python
def radar_factory(num_vars, frame="circle"):
    """Create a radar chart with `num_vars` axes.

    Parameters
    ----------
    num_vars : int
        Number of variables for the radar chart.
    frame : {"circle" | "polygon"}
        Shape of frame surrounding axes.
    """
    theta = np.linspace(0, 2 * np.pi, num_vars, endpoint=False)
    theta += np.pi / 2  # rotate so first axis points up

    def draw_poly_patch(self):
        verts = unit_poly_verts(theta)
        return plt.Polygon(verts, closed=True, edgecolor="#4E5E6E")

    def draw_circle_patch(self):
        return plt.Circle((0.5, 0.5), 0.5)

    patch_dict = {"polygon": draw_poly_patch, "circle": draw_circle_patch}
    if frame not in patch_dict:
        raise ValueError("unknown value for `frame`: %s" % frame)

    class RadarAxes(PolarAxes):
        name = "radar"
        RESOLUTION = 1
        draw_patch = patch_dict[frame]

        def fill(self, *args, **kwargs):
            closed = kwargs.pop("closed", True)
            return super(RadarAxes, self).fill(closed=closed, *args, **kwargs)

        def plot(self, *args, **kwargs):
            lines = super(RadarAxes, self).plot(*args, **kwargs)
            for line in lines:
                self._close_line(line)

        def _close_line(self, line):
            x, y = line.get_data()
            if x[0] != x[-1]:
                x = np.concatenate((x, [x[0]]))
                y = np.concatenate((y, [y[0]]))
                line.set_data(x, y)

        def set_varlabels(self, labels):
            self.set_thetagrids(np.degrees(theta), labels)

        def _gen_axes_patch(self):
            return self.draw_patch()

        def _gen_axes_spines(self):
            if frame == "circle":
                return PolarAxes._gen_axes_spines(self)
            spine_type = "circle"
            verts = unit_poly_verts(theta)
            verts.append(verts[0])
            path = Path(verts)
            spine = Spine(self, spine_type, path)  # wraps the polygon path as the axes border
            spine.set_transform(self.transAxes)
            return {"polar": spine}

    register_projection(RadarAxes)
    return theta


def unit_poly_verts(theta):
    """Return vertices of polygon for subplot axes.

    The polygon is circumscribed by a unit circle centered at (0.5, 0.5).
    """
    x0, y0, r = 0.5, 0.5, 0.5
    return [(r * np.cos(t) + x0, r * np.sin(t) + y0) for t in theta]


# --- Sample data: two teams rated across six dimensions (scale 0-10) ---
spoke_labels = ["Collaboration", "Innovation", "Resilience",
                "Communication", "Adaptability", "Leadership"]

team_a_scores = [7.2, 8.5, 6.1, 7.8, 9.0, 6.5]
team_b_scores = [6.0, 7.0, 6.5, 7.0, 7.5, 8.0]

N = len(spoke_labels)
theta = radar_factory(N, frame="polygon")

fig = plt.figure(figsize=(8, 8), dpi=120)
ax = fig.add_subplot(111, projection="radar")
plt.rgrids([2, 4, 6, 8, 10])

color_a = "#275170"
color_b = "#F77F03"

ax.plot(theta, team_a_scores, color=color_a, linewidth=1.5)
ax.fill(theta, team_a_scores, facecolor=color_a, alpha=0.25)

ax.plot(theta, team_b_scores, color=color_b, linewidth=1.5)
ax.fill(theta, team_b_scores, facecolor=color_b, alpha=0.25)

ax.set_varlabels(spoke_labels)
ax.set_ylim([0, 10])

patch_a = mpatches.Patch(color=color_a, alpha=0.85, label="Team A")
patch_b = mpatches.Patch(color=color_b, alpha=0.85, label="Team B")
lgd = plt.legend(handles=[patch_a, patch_b], loc=(0.85, 1.05), frameon=True)

tit = plt.suptitle("Performance Radar - Team Comparison", fontsize=14,
                   fontweight="bold", y=1.02)

plt.tight_layout()
plt.savefig("images/radar_plot_advanced.png",
            bbox_extra_artists=(lgd, tit), bbox_inches="tight",
            facecolor="white", edgecolor="none")
plt.show()
````
